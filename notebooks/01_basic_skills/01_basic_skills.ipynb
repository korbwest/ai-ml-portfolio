# Визуализация

```python
import matplotlib.pyplot as plt

data = df[["Model", "Fuel type"]].copy()
print(data)
```

## 1.Линейная диаграмма (plot)

```python
# Средняя цена по годам
plt.figure(figsize=(10, 5))
df.groupby('Prod. year')['Price'].mean().plot() # берем среднее по каждой группе и строим диаграмму
plt.title('Средняя цена по годам')
plt.xlabel('Год')
plt.ylabel('Цена')
plt.grid(True)
plt.show()
```

Вывод: распределение средней цены по годам, заметно, что пик был в промежутке (1940г. - 1950г.) и ближе к 2000г.

## 2. Диаграмма рассеяния (scatter)

```python
plt.figure(figsize=(15, 5))
df[0:200].plot.scatter(x="Mileage", y="Price")
plt.title('Зависимость цены от пробега')
plt.xlabel('Пробег')
plt.ylabel('Цена')
plt.show()
```

```python
Вывод: Чем меньше пробег автомобиля, тем выше цена на автомобиль. Данные по первым 200-ам записям
```

## 3. Круговая диаграмма (pie).

```python
# Доля производителей
plt.figure(figsize=(8, 8))
df['Manufacturer'].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')
plt.title('Доля производителей')
plt.ylabel('')
plt.show()
```

Вывод: наибольшая доля приходится на: Hyundai, далее: Toyota, Mercedes-benz
